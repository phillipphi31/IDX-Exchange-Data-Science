```python
list202510 = pd.read_csv('/Users/phillipphi/Desktop/IDX Exchange/Listing csv/CRMLSListing202510.csv')
list202511 = pd.read_csv('/Users/phillipphi/Desktop/IDX Exchange/Listing csv/CRMLSListing202511.csv')
list202512 = pd.read_csv('/Users/phillipphi/Desktop/IDX Exchange/Listing csv/CRMLSListing202512.csv')
list202601 = pd.read_csv('/Users/phillipphi/Desktop/IDX Exchange/Listing csv/CRMLSListing202601.csv')
list202602 = pd.read_csv('/Users/phillipphi/Desktop/IDX Exchange/Listing csv/CRMLSListing202602.csv')
list202603 = pd.read_csv('/Users/phillipphi/Desktop/IDX Exchange/Listing csv/CRMLSListing202603.csv')

listings = pd.concat([list202510, list202511, list202512, list202601, list202602, list202603], ignore_index=True)
```

```python
sold202510 = pd.read_csv('/Users/phillipphi/Desktop/IDX Exchange/Sold csv/CRMLSSold202510.csv')
sold202511 = pd.read_csv('/Users/phillipphi/Desktop/IDX Exchange/Sold csv/CRMLSSold202511.csv')
sold202512 = pd.read_csv('/Users/phillipphi/Desktop/IDX Exchange/Sold csv/CRMLSSold202512.csv')
sold202601 = pd.read_csv('/Users/phillipphi/Desktop/IDX Exchange/Sold csv/CRMLSSold202601.csv')
sold202602 = pd.read_csv('/Users/phillipphi/Desktop/IDX Exchange/Sold csv/CRMLSSold202602.csv')
sold202603 = pd.read_csv('/Users/phillipphi/Desktop/IDX Exchange/Sold csv/CRMLSSold202603.csv')
sold202604 = pd.read_csv('/Users/phillipphi/Desktop/IDX Exchange/Sold csv/CRMLSSold202604.csv')

sold = pd.concat([sold202510, sold202511, sold202512, sold202601, sold202602, sold202603], ignore_index=True)
```

```python
listings = listings.loc[(listings["PropertyType"] == "Residential") & (listings["PropertySubType"] == "SingleFamilyResidence")]
listings.head()
```

```python
sold = sold.loc[(sold["PropertyType"] == "Residential") & (sold["PropertySubType"] == "SingleFamilyResidence")]
sold.head()
```

```python
target_vars = housing[[ "ClosePrice_y", "LivingArea_y", "BedroomsTotal_y", "BathroomsTotalInteger_y", "LotSizeSquareFeet_y"]]
target_vars
```
